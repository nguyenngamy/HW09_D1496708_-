```python
# 1. 建立固定活動名稱與每人收費標準（保留原始資料，不直接覆蓋）
activity_names = ("桌遊體驗", "電影欣賞")  # tuple：保存固定的兩個活動名稱
unit_fees = {
    "桌遊體驗": 120,  # 每人 120 元
    "電影欣賞": 80    # 每人 80 元
}  # dict：以活動名稱為鍵查詢每人費用


def run_activity_test(test_label, board_raw, movie_raw):
    # 從 tuple 讀取兩個活動名稱
    board_name, movie_name = activity_names

    # 1. 整理文字名單：移除半形空白、轉小寫，再依逗號拆成 list，保留原本順序與重複項目
    board_clean = board_raw.replace(" ", "").lower()
    movie_clean = movie_raw.replace(" ", "").lower()

    # 若為空字串則建立空 list，避免直接 split(",") 產生 [""]
    board_list = board_clean.split(",") if board_clean != "" else []
    movie_list = movie_clean.split(",") if movie_clean != "" else []

    # 2. 轉換為 set 進行去重與集合運算比較名單
    board_set = set(board_list)               # 桌遊體驗去重集合
    movie_set = set(movie_list)               # 電影欣賞去重集合

    common_set = board_set & movie_set        # 共同參加者：兩場都有參加（交集）
    all_set = board_set | movie_set           # 全部參加者：至少參加一場（聯集）
    board_only_set = board_set - movie_set    # 只參加桌遊者：桌遊有、電影沒有（差集）
    movie_only_set = movie_set - board_set    # 只參加電影者：電影有、桌遊沒有（差集）

    # 依小寫英文姓名字母順序排序後再顯示
    board_sorted = sorted(board_set)
    movie_sorted = sorted(movie_set)
    common_sorted = sorted(common_set)
    all_sorted = sorted(all_set)
    board_only_sorted = sorted(board_only_set)
    movie_only_sorted = sorted(movie_only_set)

    # 3. 將「全部參加者」排序後，使用切片 [:2] 取得前兩位作為「通知名單預覽」
    preview_list = all_sorted[:2]

    # 4. 計算人數與應收費用，存入 summary 字典（金額保留為數值型別）
    board_count = len(board_set)
    movie_count = len(movie_set)
    common_count = len(common_set)
    total_unique_count = len(all_set)
    total_registrations = board_count + movie_count  # 報名人次：兩場各自去重後人數相加

    board_fee = board_count * unit_fees[board_name]
    movie_fee = movie_count * unit_fees[movie_name]
    total_fee = board_fee + movie_fee

    summary = {
        "桌遊人數": board_count,
        "電影人數": movie_count,
        "共同人數": common_count,
        "總不重複人數": total_unique_count,
        "報名人次": total_registrations,
        "桌遊應收": board_fee,
        "電影應收": movie_fee,
        "總應收": total_fee
    }

    # 5. 使用 print() 與 f-string 輸出完整摘要
    print(f"================ {test_label} ================")
    print(f"原始名單 -> {board_name}：{board_raw!r}｜{movie_name}：{movie_raw!r}")
    print(f"活動與費用 -> {board_name}：每人 {unit_fees[board_name]} 元｜{movie_name}：每人 {unit_fees[movie_name]} 元")
    print(f"整理後串列（保留重複）-> {board_name}：{board_list}｜{movie_name}：{movie_list}")
    print(f"各場去重排序名單 -> {board_name}（{summary['桌遊人數']} 人）：{board_sorted}")
    print(f"各場去重排序名單 -> {movie_name}（{summary['電影人數']} 人）：{movie_sorted}")
    print(f"共同參加者（{summary['共同人數']} 人）：{common_sorted}")
    print(f"全部參加者（{summary['總不重複人數']} 人）：{all_sorted}")
    print(f"只參加{board_name}：{board_only_sorted}")
    print(f"只參加{movie_name}：{movie_only_sorted}")
    print(f"通知名單預覽（前兩位）：{preview_list}")
    print(
        f"人數摘要 -> 總不重複人數：{summary['總不重複人數']} 人｜"
        f"報名人次：{summary['報名人次']} 人次"
    )
    print(
        f"費用摘要 -> {board_name}應收：{summary['桌遊應收']:,} 元｜"
        f"{movie_name}應收：{summary['電影應收']:,} 元｜"
        f"總應收：{summary['總應收']:,} 元\n"
    )


# --- 測試一（保留原始文字不覆蓋）---
board_raw_1 = " Amy, Ben, amy, Cora "
movie_raw_1 = " ben, Dora, ERIC, ben "
run_activity_test("測試一", board_raw_1, movie_raw_1)

# --- 測試二 ---
board_raw_2 = " Leo, mia, LEO, nina, Mia "
movie_raw_2 = " mia, Oscar, leo, tina "
run_activity_test("測試二", board_raw_2, movie_raw_2)

# --- 測試三（空名單）---
board_raw_3 = ""
movie_raw_3 = ""
run_activity_test("測試三（空名單）", board_raw_3, movie_raw_3)
```
